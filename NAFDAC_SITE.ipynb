From the above, it appears NAFDAC numbers often follow a pattern like `XX-XXXX` or `XX-XXXXL`, or `XX-XXXXF`. Some also have additional characters. We'll define a flexible regex to capture these variations.

```python
import re
import pandas as pd

def sanitize_nafdac_number(nafdac_input):
    if pd.isna(nafdac_input):
        return None

    s = str(nafdac_input).strip().upper() # Convert to uppercase early for consistent matching

    # Handle known non-NAFDAC placeholders
    if not s or s in ['NA', 'NOT AVAILABLE YET', 'N/A']:
        return None

    # Remove all spaces and normalize various separators ('-', '–', '/') to a single hyphen
    s = re.sub(r'\s', '', s)
    s = re.sub(r'[\-\u2013/]', '-', s)

    # Define valid prefix letters for categories (A, B, C, D, E, F, W, V, COS, VET)
    category_prefix_regex = r'(A|B|C|D|E|F|W|V|COS|VET)'
    # Define valid suffix letters for product types (L, T, C, P, O, K, N)
    product_type_suffix_regex = r'(L|T|C|P|O|K|N)'

    # Attempt to match the structured NAFDAC patterns (e.g., A4-2345T, 04-5112)
    # The regex is built using concatenation of raw strings to avoid SyntaxWarning
    # Group structure for the main_pattern_match_pattern (capturing groups only):
    # 1: (category_prefix_regex) - e.g., 'A', 'COS'
    # 2: (\d{1,3}|[A-Z]) - batch identifier (1-3 digits or a single letter), optional
    # 3: (\d{2}) - two-digit prefix (alternative) - e.g., '04'
    # 4: (\d{4,6}) - 4-6 serial digits - e.g., '4567'
    # 5: ([A-Z]*) - Optional raw suffix - e.g., 'T' or ''
    main_pattern_match_pattern = (
        r'^(?:' # Start of non-capturing alternation group for prefixes
        + category_prefix_regex + r'(\d{1,3}|[A-Z])?'  # Branch 1: (Category) (Batch Identifier, optional)
        r'|'
        r'(\d{2})'                               # Branch 2: (Two-digit Numeric Prefix)
        r')-'
        r'(\d{4,6})'
        r'([A-Z]*)'
        r'$'
    )
    main_pattern_match = re.match(main_pattern_match_pattern, s)

    if main_pattern_match:
        prefix_part = ''
        serial_digits = ''
        raw_suffix = ''

        # Determine which prefix type matched based on group existence
        if main_pattern_match.group(1): # Category prefix (e.g., A, COS) matched (group 1 is the category itself)
            prefix = main_pattern_match.group(1)
            batch_identifier = main_pattern_match.group(2)
            prefix_part = prefix + (batch_identifier if batch_identifier else '')
            serial_digits = main_pattern_match.group(4)
            raw_suffix = main_pattern_match.group(5)
        elif main_pattern_match.group(3): # Two-digit prefix (e.g., 04, 11) matched
            prefix_part = main_pattern_match.group(3)
            serial_digits = main_pattern_match.group(4)
            raw_suffix = main_pattern_match.group(5)

        final_suffix = ''
        # If there's a raw suffix, check if it's a single valid suffix letter
        if raw_suffix and len(raw_suffix) == 1 and re.match(r'^' + product_type_suffix_regex + r'$', raw_suffix):
            final_suffix = raw_suffix

        if prefix_part and serial_digits: # Ensure we have both parts for a valid structure
             return f'{prefix_part}-{serial_digits}{final_suffix}'
        else: # If parts are missing, it implies a partial or invalid match for this path
            return None

    # Fallback for purely numeric strings (5+ digits), for 'older products' that might not follow new structure
    match_digits_only = re.match(r'^(\d{5,})$', s)
    if match_digits_only:
        return match_digits_only.group(1)

    return None

# The `nafdac_validation_regex` will strictly validate the output of `sanitize_nafdac_number`.
# It combines three patterns:
# 1. CategoryPrefix (A, B, COS, etc.) + optional (1-3 BatchDigits or a single letter) - 4-6 SerialDigits + OptionalSingleValidSuffix
# 2. TwoDigits as Prefix - 4-6 SerialDigits + OptionalSingleValidSuffix
# 3. Purely 5+ Digits (for older numbers)
nafdac_validation_regex = (
    r'^(?:'
    r'(?:A|B|C|D|E|F|W|V|COS|VET)(?:\d{1,3}|[A-Z])?-\d{4,6}(?:L|T|C|P|O|K|N)?' # Updated batch part
    r'|'
    r'\d{2}-\d{4,6}(?:L|T|C|P|O|K|N)?'
    r'|'
    r'\d{5,}'
    r')$'
)

def validate_nafdac_format(sanitized_nafdac_number):
    # If sanitization returned None, it's inherently invalid
    if sanitized_nafdac_number is None:
        return False
    if not isinstance(sanitized_nafdac_number, str):
        return False
    return bool(re.match(nafdac_validation_regex, sanitized_nafdac_number))

# Apply sanitization first to create a new, cleaned NAFDAC column
df['NAFDAC_cleaned'] = df['NAFDAC'].apply(sanitize_nafdac_number)

# Then apply the validation to the cleaned NAFDAC numbers
df['is_nafdac_format_valid'] = df['NAFDAC_cleaned'].apply(validate_nafdac_format)

print('\nNAFDAC format validation results (after refined sanitization):')
display(df['is_nafdac_format_valid'].value_counts())

# Display some examples of original NAFDAC values that led to invalid format after sanitization
print('\nExamples of original NAFDAC values that are still invalid after sanitization:')
# Filter the original 'NAFDAC' column where the cleaned one was deemed invalid
display(df[~df['is_nafdac_format_valid']][['NAFDAC', 'NAFDAC_cleaned']].head(10))
```

### Step 3: Local High-Risk Database Lookup

This step involves checking validated NAFDAC numbers against a local database of known high-risk or counterfeit product NAFDACs. For demonstration purposes, we'll create a simulated high-risk list.

```python
# Simulate a local high-risk NAFDAC database (in a real scenario, this would be a loaded dataset)
high_risk_nafdacs = {
    'A123-4567T', # Example high-risk NAFDAC
    'B01-2023L',  # Another example
    '04-5112',    # Example of a numeric NAFDAC in high-risk
    'A11A-1234',  # An invalid NAFDAC that might be in a raw list, but should be caught by validation
    'B1-7187'     # Example of the newly validated format
}

def lookup_high_risk(nafdac_number):
    """
    Checks if a given NAFDAC number is present in the high-risk database.

    Args:
        nafdac_number (str): The sanitized NAFDAC number to check.

    Returns:
        bool: True if the NAFDAC is high-risk, False otherwise.
    """
    if nafdac_number is None:
        return False # A non-valid NAFDAC cannot be high-risk in this context
    return nafdac_number in high_risk_nafdacs

# Apply the high-risk lookup to the cleaned and validated NAFDACs
df['is_high_risk'] = df['NAFDAC_cleaned'].apply(lookup_high_risk)

print('\nHigh-Risk NAFDAC Lookup Results:')
display(df['is_high_risk'].value_counts())

print('\nExamples of NAFDACs identified as high-risk:')
display(df[df['is_high_risk']][['NAFDAC', 'NAFDAC_cleaned', 'is_nafdac_format_valid', 'is_high_risk']].head())
```

### Step 4: Scale to Real-World Verification (Simulated API and Metadata Check)

To address the dynamic interface with the NAFDAC official database, we will simulate this behavior. We'll create a mock Greenbook database containing NAFDAC numbers and their associated metadata. Then, we'll modify our `nafdac_verification_pipeline` to query this simulated database and perform metadata matching to distinguish between 'Genuine', 'Cloned', and 'Counterfeit (Missing from DB)' products.

```python
# --- 1. Simulate NAFDAC Greenbook Database (API Response) ---
# This dictionary acts as our mock NAFDAC Greenbook database, mapping sanitized NAFDACs to product metadata.
# Since the original data was scraped from the NAFDAC Greenbook, we will populate this mock DB
# with all the NAFDAC_cleaned entries and their product_name from the DataFrame.

# Initialize with some specific examples (e.g., high-risk ones that we want to test)
nafdac_greenbook_db_mock = {
    'A123-4567T': {'product_name': 'Sample Drug A'}, # Genuine example
    'B01-2023L': {'product_name': 'Another Product'}, # Genuine example
    'C11-0003P': {'product_name': 'Vitamin C'},     # Genuine example
    '04-5112': {'product_name': 'Fake Product Name'}, # This was high-risk, but if found, imagine a mismatch
    'A11-0004K': {'product_name': 'Genuine Item'}, # Will be Genuine
    'VET-0005N': {'product_name': 'Animal Health Product'},
    'B1-7187': {'product_name': 'Fake Drug B'}
}

# Populate the mock database with all NAFDAC_cleaned entries from the df
# For simplicity, if a NAFDAC_cleaned appears multiple times, we'll take the first product_name.
for index, row in df.iterrows():
    nafdac_c = row['NAFDAC_cleaned']
    prod_name = row['product_name']
    if nafdac_c and nafdac_c not in nafdac_greenbook_db_mock: # Avoid overwriting our specific test cases
        nafdac_greenbook_db_mock[nafdac_c] = {'product_name': prod_name}

def simulate_nafdac_api_lookup(sanitized_nafdac_number):
    """
    Simulates an API call to the NAFDAC Greenbook database.

    Args:
        sanitized_nafdac_number (str): The sanitized NAFDAC number.

    Returns:
        dict: Metadata if found, else None.
    """
    return nafdac_greenbook_db_mock.get(sanitized_nafdac_number)

# --- 2. Prepare Local Product Metadata for Comparison ---
# Create a dictionary from our DataFrame mapping NAFDAC_cleaned to its product_name.
# This will serve as our 'local' product metadata.

# Handle potential duplicates in NAFDAC_cleaned by taking the first product name
# This is a simplification; in a real scenario, you'd define how to handle multiple products per NAFDAC.
local_nafdac_metadata = df.groupby('NAFDAC_cleaned')['product_name'].apply(lambda x: x.iloc[0]).to_dict()

print("Simulated NAFDAC Greenbook DB and Local Metadata prepared with DataFrame data.")
```

### Step 5: Integrated Pipeline Engine

Now, we'll combine the `sanitize_nafdac_number`, `validate_nafdac_format`, and `lookup_high_risk` functions into a single pipeline function that provides a comprehensive verdict for each NAFDAC number. The verdicts will be:

*   **Authentic**: The NAFDAC number is well-formatted and not found in the high-risk database.
*   **Counterfeit (High-Risk)**: The NAFDAC number is well-formatted and found in the high-risk database.
*   **Invalid Format**: The NAFDAC number does not conform to the expected format after sanitization.
*   **Unknown**: The input was empty, an unidentifiable placeholder, or could not be processed.

```python
import re

def normalize_product_name(name):
    """
    Normalizes product names for more flexible comparison.
    Converts to lowercase, removes common trailing special characters, and non-alphanumeric chars.
    """
    if not name:
        return ""
    name = str(name).lower()
    name = name.strip()
    # Remove common trailing special characters and symbols that are often inconsistent
    name = re.sub(r'[\s*#$]+$', '', name)
    # Remove non-alphanumeric characters, but keep spaces
    name = re.sub(r'[^a-z0-9\s]', '', name)
    # Reduce multiple spaces to single spaces
    name = re.sub(r'\s+', ' ', name)
    return name

def nafdac_verification_pipeline(nafdac_input, local_product_name=None):
    """
    Integrates sanitization, format validation, high-risk lookup, and simulated Greenbook API lookup
    with metadata comparison to provide a comprehensive verdict for a NAFDAC number.
    Verdicts are simplified to 'Genuine' or 'Suspicious' as per user request.

    Args:
        nafdac_input (str): The raw NAFDAC number string.
        local_product_name (str, optional): The product name from our local dataset for metadata comparison.

    Returns:
        str: A simplified verdict ('Genuine' or 'Suspicious').
    """
    sanitized = sanitize_nafdac_number(nafdac_input)

    if sanitized is None:
        return 'Suspicious' # Unknown/uncleanable input

    is_valid_format = validate_nafdac_format(sanitized)

    if not is_valid_format:
        return 'Suspicious' # Invalid format

    is_high_risk = lookup_high_risk(sanitized)

    if is_high_risk:
        return 'Suspicious' # High-risk

    # --- Simulated Greenbook API Lookup ---
    greenbook_entry = simulate_nafdac_api_lookup(sanitized)

    if greenbook_entry is None:
        # Rule: "Counterfeit" if missing completely from the database (now 'Suspicious')
        return 'Suspicious' # Not found in Greenbook (simulated)
    else:
        # If Greenbook entry is found, and it passed previous checks (sanitized, valid format, not high-risk),
        # then we consider it Genuine. Metadata comparison will not downgrade to 'Suspicious'.
        return 'Genuine'

# Apply the updated pipeline to the original NAFDAC column and include 'product_name' for metadata comparison
df['nafdac_verdict_simplified'] = df.apply(lambda row: nafdac_verification_pipeline(
    row['NAFDAC'],
    local_product_name=local_nafdac_metadata.get(row['NAFDAC_cleaned']) # Pass product name from our local metadata map
), axis=1)

print('\nComprehensive NAFDAC Verification Pipeline Results (Simplified Verdicts):')
display(df['nafdac_verdict_simplified'].value_counts())

print('\nExamples of each simplified verdict category:')
for verdict in df['nafdac_verdict_simplified'].unique():
    print(f'\nVerdict: {verdict}')
    display(df[df['nafdac_verdict_simplified'] == verdict][['NAFDAC', 'product_name', 'NAFDAC_cleaned', 'is_nafdac_format_valid', 'is_high_risk', 'nafdac_verdict_simplified']].head(3))
```

### Step 6: Interactive NAFDAC Verification System

This interactive tool allows you to input any NAFDAC number (and optionally a product name for metadata comparison) to instantly get a 'Genuine' or 'Suspicious' verdict from our comprehensive pipeline.

```python
from ipywidgets import interact, Text, Button, VBox, Output
from IPython.display import display

def interactive_nafdac_verifier(nafdac_input, product_name_input=""): # Default product_name_input to empty string
    # Use the pipeline to get the verdict
    verdict = nafdac_verification_pipeline(nafdac_input, product_name_input)

    # Also run individual checks for detailed feedback
    sanitized = sanitize_nafdac_number(nafdac_input)
    is_valid = validate_nafdac_format(sanitized) if sanitized else False
    is_risk = lookup_high_risk(sanitized) if sanitized else False

    if sanitized:
        greenbook_check = simulate_nafdac_api_lookup(sanitized)
        greenbook_status = 'Found' if greenbook_check else 'Not Found (Simulated)'
        greenbook_prod_name = greenbook_check.get('product_name') if greenbook_check else 'N/A'
    else:
        greenbook_status = 'N/A (Not sanitized)'
        greenbook_prod_name = 'N/A'

    # Display the results
    with output_area:
        output_area.clear_output(wait=True)
        print(f"--- Verification Result for '{nafdac_input}' ---")
        print(f"Raw Input: {nafdac_input}")
        print(f"Sanitized NAFDAC: {sanitized if sanitized else 'N/A'}")
        print(f"Format Valid: {is_valid}")
        print(f"High-Risk: {is_risk}")
        print(f"Greenbook Status: {greenbook_status}")
        print(f"Greenbook Product Name: {greenbook_prod_name}")
        print(f"Local Product Name Provided: {product_name_input if product_name_input else 'None'}")
        print(f"\nFINAL VERDICT: {verdict}")
        if verdict == 'Genuine':
            print("This NAFDAC number appears authentic and matches Greenbook data.")
        else:
            print("This NAFDAC number is suspicious. Review details above for reasons.")
        print("-------------------------------------------")

# Create interactive widgets
nafdac_text = Text(description='NAFDAC Number:', placeholder='e.g., A7-0001L, 04-5112')
product_name_text = Text(description='Product Name (Optional):', placeholder='e.g., Sample Drug A')
verify_button = Button(description='Verify NAFDAC')
output_area = Output()

def on_button_click(b):
    interactive_nafdac_verifier(nafdac_text.value, product_name_text.value)

verify_button.on_click(on_button_click)

# Display the widgets
print("Please enter a NAFDAC number and optionally a product name to verify:")
display(VBox([nafdac_text, product_name_text, verify_button, output_area]))
```

```python
%%writefile app.py
import streamlit as st
import joblib
import re
import pandas as pd # Although not directly used in the app, it's part of the original functions

# --- Load the saved data structures ---
# Ensure these files are in the /content/ directory or provide full paths
greenbook_db_filename = 'nafdac_greenbook_db_mock.joblib'
local_metadata_filename = 'local_nafdac_metadata.joblib'

try:
    nafdac_greenbook_db_mock = joblib.load(greenbook_db_filename)
    local_nafdac_metadata = joblib.load(local_metadata_filename)
    st.success("Mock Greenbook DB and Local Metadata loaded successfully!")
except FileNotFoundError:
    st.error("Error: NAFDAC data files not found. Please ensure 'nafdac_greenbook_db_mock.joblib' and 'local_nafdac_metadata.joblib' are in the current directory.")
    st.stop() # Stop the app if data is not loaded

# --- Define NAFDAC Verification Functions (Copied from previous cells) ---

def sanitize_nafdac_number(nafdac_input):
    if pd.isna(nafdac_input):
        return None

    s = str(nafdac_input).strip().upper() # Convert to uppercase early for consistent matching

    # Handle known non-NAFDAC placeholders
    if not s or s in ['NA', 'NOT AVAILABLE YET', 'N/A']:
        return None

    # Remove all spaces and normalize various separators ('-', '–', '/') to a single hyphen
    s = re.sub(r'\s', '', s)
    s = re.sub(r'[\-\u2013/]', '-', s)

    # Define valid prefix letters for categories (A, B, C, D, E, F, W, V, COS, VET)
    category_prefix_regex = r'(A|B|C|D|E|F|W|V|COS|VET)'
    # Define valid suffix letters for product types (L, T, C, P, O, K, N)
    product_type_suffix_regex = r'(L|T|C|P|O|K|N)'

    # Attempt to match the structured NAFDAC patterns (e.g., A4-2345T, 04-5112)
    # The regex is built using concatenation of raw strings to avoid SyntaxWarning
    # Group structure for the main_pattern_match_pattern (capturing groups only):
    # 1: (category_prefix_regex) - e.g., 'A', 'COS'
    # 2: (\d{1,3}|[A-Z]) - batch identifier (1-3 digits or a single letter), optional
    # 3: (\d{2}) - two-digit prefix (alternative) - e.g., '04'
    # 4: (\d{4,6}) - 4-6 serial digits - e.g., '4567'
    # 5: ([A-Z]*) - Optional raw suffix - e.g., 'T' or ''
    main_pattern_match_pattern = (
        r'^(?:' # Start of non-capturing alternation group for prefixes
        + category_prefix_regex + r'(\d{1,3}|[A-Z])?'  # Branch 1: (Category) (Batch Identifier, optional)
        r'|'
        r'(\d{2})'                               # Branch 2: (Two-digit Numeric Prefix)
        r')-'
        r'(\d{4,6})'
        r'([A-Z]*)'
        r'$'
    )
    main_pattern_match = re.match(main_pattern_match_pattern, s)

    if main_pattern_match:
        prefix_part = ''
        serial_digits = ''
        raw_suffix = ''

        # Determine which prefix type matched based on group existence
        if main_pattern_match.group(1): # Category prefix (e.g., A, COS) matched (group 1 is the category itself)
            prefix = main_pattern_match.group(1)
            batch_identifier = main_pattern_match.group(2)
            prefix_part = prefix + (batch_identifier if batch_identifier else '')
            serial_digits = main_pattern_match.group(4)
            raw_suffix = main_pattern_match.group(5)
        elif main_pattern_match.group(3): # Two-digit prefix (e.g., 04, 11) matched
            prefix_part = main_pattern_match.group(3)
            serial_digits = main_pattern_match.group(4)
            raw_suffix = main_pattern_match.group(5)

        final_suffix = ''
        # If there's a raw suffix, check if it's a single valid suffix letter
        if raw_suffix and len(raw_suffix) == 1 and re.match(r'^' + product_type_suffix_regex + r'$', raw_suffix):
            final_suffix = raw_suffix

        if prefix_part and serial_digits: # Ensure we have both parts for a valid structure
             return f'{prefix_part}-{serial_digits}{final_suffix}'
        else: # If parts are missing, it implies a partial or invalid match for this path
            return None

    # Fallback for purely numeric strings (5+ digits), for 'older products' that might not follow new structure
    match_digits_only = re.match(r'^(\d{5,})$', s)
    if match_digits_only:
        return match_digits_only.group(1)

    return None

# The `nafdac_validation_regex` will strictly validate the output of `sanitize_nafdac_number`.
# It combines three patterns:
# 1. CategoryPrefix (A, B, COS, etc.) + optional (1-3 BatchDigits or a single letter) - 4-6 SerialDigits + OptionalSingleValidSuffix
# 2. TwoDigits as Prefix - 4-6 SerialDigits + OptionalSingleValidSuffix
# 3. Purely 5+ Digits (for older numbers)
nafdac_validation_regex = (
    r'^(?:'
    r'(?:A|B|C|D|E|F|W|V|COS|VET)(?:\d{1,3}|[A-Z])?-\d{4,6}(?:L|T|C|P|O|K|N)?' # Updated batch part
    r'|'
    r'\d{2}-\d{4,6}(?:L|T|C|P|O|K|N)?'
    r'|'
    r'\d{5,}'
    r')$'
)

def validate_nafdac_format(sanitized_nafdac_number):
    # If sanitization returned None, it's inherently invalid
    if sanitized_nafdac_number is None:
        return False
    if not isinstance(sanitized_nafdac_number, str):
        return False
    return bool(re.match(nafdac_validation_regex, sanitized_nafdac_number))

# Simulate a local high-risk NAFDAC database
high_risk_nafdacs = {
    'A123-4567T',
    'B01-2023L',
    '04-5112',
    'B1-7187'
}

def lookup_high_risk(nafdac_number):
    if nafdac_number is None:
        return False
    return nafdac_number in high_risk_nafdacs

def simulate_nafdac_api_lookup(sanitized_nafdac_number):
    return nafdac_greenbook_db_mock.get(sanitized_nafdac_number)

def normalize_product_name(name):
    if not name:
        return ""
    name = str(name).lower()
    name = name.strip()
    name = re.sub(r'[\s*#$]+$', '', name)
    name = re.sub(r'[^a-z0-9\s]', '', name)
    name = re.sub(r'\s+', ' ', name)
    return name

def nafdac_verification_pipeline(nafdac_input, local_product_name=None):
    sanitized = sanitize_nafdac_number(nafdac_input)

    if sanitized is None:
        return 'Suspicious' # Unknown/uncleanable input

    is_valid_format = validate_nafdac_format(sanitized)

    if not is_valid_format:
        return 'Suspicious' # Invalid format

    is_high_risk = lookup_high_risk(sanitized)

    if is_high_risk:
        return 'Suspicious' # High-risk

    greenbook_entry = simulate_nafdac_api_lookup(sanitized)

    if greenbook_entry is None:
        return 'Suspicious' # Not found in Greenbook (simulated)
    else:
        # Check if a product name was provided for comparison
        if local_product_name:
            greenbook_product_name = greenbook_entry.get('product_name')
            if greenbook_product_name:
                normalized_provided_name = normalize_product_name(local_product_name)
                normalized_greenbook_name = normalize_product_name(greenbook_product_name)
                # Simple substring check (can be improved with fuzzy matching if needed)
                if normalized_provided_name in normalized_greenbook_name or \
                   normalized_greenbook_name in normalized_provided_name:
                    return 'Genuine'
                else:
                    # NAFDAC found, but product name mismatch -> Cloned/Suspicious
                    return 'Suspicious' # NAFDAC found, but name mismatch
            else:
                # NAFDAC found, but no product name in greenbook entry for comparison
                return 'Genuine' # Assume genuine if NAFDAC exists and no name to compare
        else:
            # No product name provided, NAFDAC found and passed other checks
            return 'Genuine'

# --- Streamlit UI Layout ---
st.set_page_config(layout="centered", page_title="NAFDAC Product Verification")
st.title("🇳🇬 NAFDAC Product Verification System")
st.markdown("Enter a NAFDAC number and optionally a product name to get a real-time verification verdict.")

nafdac_input = st.text_input("NAFDAC Number", placeholder="e.g., A7-0001L, 04-5112")
product_name_input = st.text_input("Product Name (Optional)", placeholder="e.g., Sample Drug A")

if st.button("Verify NAFDAC"):
    if nafdac_input:
        # Call your existing pipeline function
        # For the Streamlit app, we'll get a detailed output string from a helper function
        # that wraps the pipeline and individual checks, similar to the Gradio approach.

        verdict = nafdac_verification_pipeline(nafdac_input, product_name_input)

        # Prepare detailed output for display
        sanitized = sanitize_nafdac_number(nafdac_input)
        is_valid = validate_nafdac_format(sanitized) if sanitized else False
        is_risk = lookup_high_risk(sanitized) if sanitized else False

        greenbook_status_str = 'N/A'
        greenbook_prod_name_str = 'N/A'
        name_match_status = 'N/A'

        if sanitized:
            greenbook_check = simulate_nafdac_api_lookup(sanitized)
            greenbook_status_str = 'Found' if greenbook_check else 'Not Found (Simulated)'
            greenbook_prod_name_str = greenbook_check.get('product_name') if greenbook_check else 'N/A'

            if product_name_input and greenbook_check and greenbook_check.get('product_name'):
                normalized_provided_name = normalize_product_name(product_name_input)
                normalized_greenbook_name = normalize_product_name(greenbook_check.get('product_name'))
                if normalized_provided_name in normalized_greenbook_name or \
                   normalized_greenbook_name in normalized_provided_name:
                    name_match_status = 'Matched'
                else:
                    name_match_status = 'Mismatched'
            elif product_name_input and not greenbook_check:
                 name_match_status = 'Not found in Greenbook for comparison'
            elif not product_name_input:
                 name_match_status = 'No local product name provided'

        st.subheader("Verification Details:")
        st.write(f"**Raw Input:** `{nafdac_input}`")
        st.write(f"**Sanitized NAFDAC:** `{sanitized if sanitized else 'N/A'}`")
        st.write(f"**Format Valid:** `{is_valid}`")
        st.write(f"**High-Risk Flag:** `{is_risk}`")
        st.write(f"**Greenbook DB Status:** `{greenbook_status_str}`")
        st.write(f"**Greenbook Product Name:** `{greenbook_prod_name_str}`")
        st.write(f"**Local Product Name Provided:** `{product_name_input if product_name_input else 'None'}`")
        st.write(f"**Product Name Match:** `{name_match_status}`")

        if verdict == 'Genuine':
            st.markdown(f"### FINAL VERDICT: <span style='color:green'>**{verdict}**</span>", unsafe_allow_html=True)
            st.success("This NAFDAC number appears authentic and matches Greenbook data.")
        else:
            st.markdown(f"### FINAL VERDICT: <span style='color:red'>**{verdict}**</span>", unsafe_allow_html=True)
            st.warning("This NAFDAC number is suspicious. Review details above for reasons.")
    else:
        st.warning("Please enter a NAFDAC number to verify.")
```

This was a critical phase where we employed a powerful pattern-matching algorithm: **Regular Expressions (Regex)**. We observed significant inconsistencies in how NAFDAC numbers are recorded, including variations in formatting, special characters, and placeholder text. This function cleans the raw NAFDAC input by removing extraneous characters, standardizing separators, and converting text to a consistent format. Following sanitization, a strict validation **regex** ensures that only numbers conforming to established NAFDAC patterns are processed, effectively filtering out malformed or irrelevant entries.

The NAFDAC registration numbers typically follow these template patterns:

*   **Pattern 1: Category-Based Structure**
    *   Starts with a **Category Prefix** (e.g., `A`, `B`, `C`, `D`, `E`, `F`, `W`, `V`, `COS`, `VET`).
    *   Followed by an **optional Batch Identifier** (1-3 digits or a single letter).
    *   Then a hyphen (`-`).
    *   Followed by **4 to 6 Serial Digits**.
    *   Optionally ends with a single **Product Type Suffix** (e.g., `L`, `T`, `C`, `P`, `O`, `K`, `N`).
    *   *Example: `A7-0001L`, `COS-12345`, `VETB-67890C`*

*   **Pattern 2: Two-Digit Prefix Structure**
    *   Starts with a **Two-Digit Numeric Prefix** (e.g., `04`, `11`).
    *   Then a hyphen (`-`).
    *   Followed by **4 to 6 Serial Digits**.
    *   Optionally ends with a single **Product Type Suffix**.
    *   *Example: `04-5112`, `12-9876P`*

*   **Pattern 3: Purely Numeric (Older Products)**
    *   Consists of **5 or more digits** with no letters or hyphens.
    *   This pattern accounts for older registration numbers that may not adhere to the modern structured format.
    *   *Example: `00000`, `123456`*

By validating against these comprehensive patterns, we ensure the integrity of the NAFDAC numbers entering our system.

### 4. Local High-Risk Database Lookup

*   **Counterfeit (Missing from DB):** The NAFDAC number is well-formatted but simply doesn't appear in the official Greenbook database, strongly indicating a counterfeit product.

For simplicity in the final output, we've consolidated these detailed verdicts into either 'Genuine' or 'Suspicious'.

### Project Evaluation Approach

Evaluation is central to the reliability of any system. In this project, our 'evaluation' isn't a single statistical metric like accuracy or F1-score, as we're dealing with a rule-based verification system rather than a predictive model. Instead, it's a multi-layered process where each NAFDAC number undergoes a series of stringent checks. Here's how we evaluate a NAFDAC number:



Streamlit was used to build the Interactive NAFDAC Verification System, which serves as the user-friendly front-end for the entire project. Specifically, it allowed us to create a simple web application where:

  Users can input a NAFDAC number (and optionally a product name).

### 1. NAFDAC Format Validation:

*   **Method:** This is the foundational check, implemented using **Regular Expressions**. After sanitizing the raw input (removing spaces, standardizing separators, etc.), we rigorously compare the NAFDAC number against a set of predefined NAFDAC number patterns. These patterns encapsulate the various valid structures of NAFDAC registration numbers, including category-based prefixes (e.g., `A`, `COS`), two-digit numeric prefixes (e.g., `04`), followed by serial digits and optional product type suffixes.
*   **Evaluation:** A NAFDAC number is considered 'valid' if and only if it precisely matches one of these established regular expression patterns. Any deviation immediately flags it as having an 'Invalid Format'.

### 2. High-Risk Database Lookup:

*   **Method:** We maintain a simulated local database of known high-risk NAFDAC numbers. This database represents a critical list of products that have previously been identified as counterfeit or associated with safety concerns.
*   **Evaluation:** Each sanitized and format-validated NAFDAC number is checked for its presence in this high-risk database. If a match is found, the product is immediately flagged as 'High-Risk'. This acts as an override, swiftly identifying known problematic items.

### 3. Simulated Greenbook API Lookup:

*   **Method:** This step simulates querying the official NAFDAC Greenbook. Our mock Greenbook database contains records of officially registered products and their associated metadata (like product names).
*   **Evaluation:** We check if the sanitized NAFDAC number exists within this simulated official registry. If it doesn't appear, it's a strong indicator that the product is 'Counterfeit (Missing from DB)'.

### 4. Product Name Metadata Comparison:

*   **Method:** If a NAFDAC number is found in the simulated Greenbook and a `local_product_name` is also provided by the user, we perform a comparison. Both the provided product name and the official product name from the Greenbook are first normalized (converted to lowercase, stripped of special characters, etc.). Then, we check for a substring match between the normalized names.
*   **Evaluation:** If the NAFDAC number is legitimate but the provided product name does not sufficiently match the official Greenbook entry, this suggests a potential 'Cloned' product – a legitimate NAFDAC number being used for a different product. If there's a match, it adds to the confidence of authenticity.

### Final Verdict Consolidation:

All these individual evaluations are integrated into our `nafdac_verification_pipeline`. The pipeline processes these checks sequentially:

*   If the NAFDAC is `None` after sanitization or fails format validation, it's immediately 'Suspicious'.
*   If it's found in the high-risk database, it's 'Suspicious'.
*   If it passes the above but is *not* found in the Greenbook, it's 'Suspicious'.
*   If it is found in the Greenbook, and a provided product name (if any) matches the Greenbook's product name (or no product name was provided), it is classified as 'Genuine'. If the product names mismatch, it's 'Suspicious'.

Essentially, a product must pass *all* verification layers – proper format, not high-risk, existence in the Greenbook, and matching metadata (if provided) – to be deemed 'Genuine'. Any failure along this chain results in a 'Suspicious' verdict. This layered approach ensures comprehensive scrutiny for each product.
